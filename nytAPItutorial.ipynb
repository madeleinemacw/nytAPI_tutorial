## Why use an API?
Since not all data is publicly available, firms will use APIs to organize their internal databases for external use. In the case of the New York Times (NYT), the publication provides JSON data of all articles alongside other formats. 

## How do you access an API?
To track data usage, firms will require keys. A key is your access the server containing the requested data. Consider the key a password; **never share, ESPECIALLY TO AI!**

When creating replicable code, best practice is to use a place holder like "XXXXXXX" or "INSERTKEYHERE" with commented instructions next to the line. Each key is assigned a rate limit for how many times you can access or "call" the database daily, hourly, and even by the minute. For the NYT API, the rate limits are set at 5 calls per minute with a total cap of 500 calls daily. The NYT recommends pausing for a minimum of 12 seconds between requests in order to not trigger rate limits. Higher rate limits are available upon request.

When registering for the NYT API, follow this link to set up your account: https://developer.nytimes.com/get-started 

*NOTE: Not necessarily intuitive, you'll go to "Apps" to create your first key, where you can enable all, a few, or one API available from the NYT. Most times, you only need to activate the Archive and Article Search APIs.*

````python
api_key = "INSERTKEYHERE" #insert your API key here 

nyt = NYTAPI(api_key, parse_dates=True)
````

## Available NYT APIs to Activate 
Within the NYT APIs, you can activate 5 APIs: https://developer.nytimes.com/apis

- **Archive API:** returns an array of NYT articles for a given month, going back to 1851. Its response fields are the same as the Article Search API. The Archive API is very useful if you want to build your own database of NYT article metadata. You simply pass the API the year and month and it returns a JSON object with all articles for that month. The API response size can be large (~20mb) and isn't meant to be called from the browser.

- **Article Search API:** to look up articles by keyword. You can refine your search using filters. For a list of all fields you can filter on, see documentation here - https://developer.nytimes.com/docs/articlesearch-product/1/overview

- **Books API:** provides information about books on The New York Times Best Sellers lists. Some lists are weekly (e.g. hardcover fiction) and some lists are monthly (e.g. audio fiction). New lists are published on the web on Wednesday around 7pm eastern time.

- **Most Popular API:** Provides services for getting the most popular articles on NYTimes.com based on emails, shares, or views.

- **Top Stories API:** returns an array of articles currently on the specified section (arts, business, ...)

To explore further into the individual APIs, the documentation will instruct you with examples on how to modify your queries. As an aside, RSS Feeds is technically listed within the API options, but can be called without creating a key by going directly to rss.nytimes.com and modifying the URL.
For example -  
https://rss.nytimes.com/services/xml/rss/nyt/Arts.xml
https://rss.nytimes.com/services/xml/rss/nyt/HomePage.xml

## Accessing the Top Stories API
### Using `top_stories()`
With the PYNTimes package, let's take a look at some recent history. `top_stories()` pulls the currently trending stories from that day. It does not provide historical or archival data and will change in moments of crisis news. You can specify sections the following sections: arts, automobiles, books, business, fashion, food, health, home, insider, magazine, movies, national, nyregion, obituaries, opinion, politics, realestate, science, sports, sundayreview, technology, theater, tmagazine, travel, upshot, and world.

````python
top_stories = nyt.top_stories() #call object holding all data & metadata on top articles 

top_story = top_stories[0] #narrow to first article within top_stories object, the TOP top story!

print(top_story.head()) #look within the dictionary, let's read this book! 
````

## Accessing Most Popular API 
### Using `most_viewed()`
Parsing beyond sections, you can pull the most viewed articles from a separate API. With `most_viewed()`, you can retrieve articles from yesterday, a week ago, or a month ago. Using the `days` argument, you can input either 1 to look at today and yesterday, 7 to see today and 6 days prior, or 30 for today and 29 days. Note, that the 30 days may pose problems for months with 31 days. 

````python
#Retrieve the most viewed articles for today.
most_viewed_today = nyt.most_viewed()
most_viewed_story_today = most_viewed_today[0]
print(most_viewed_story_today.head())
````

### Using `most_shared()`
Another function retrieving articles from the Most Popular API, `most_shared` pulls the top articles sent to another via email or Facebook. Like the last function, you can specify the date range. The Times does not publicly share more granular social share. Articles are ordered from most to least shares per article. 

````python
email = nyt.most_shared(days = 30, method = 'email')
print(email.head())
````

 ## Accessing Article Search API 
 ### Using `article_search()`
(https://pynytimes.michadenheijer.com/search/article-search)
With PYNTimes, you are able to query keywords using the function `article_search()`. This function searches articles that contain the set of keywords based on provided criterion. Basic parameters must include at minimum the `query` itself to return anything. You can specify total `results`, which `dates`, and list `options` for narrower parameters. The two core filters - `dates` and `options` - act like dictionaries with further subfilters to narrow returned results. 

We'll fully implement these parameters, but let's begin with the basics. Without all the fancy filters, take a look at the object called by `article_search()`. 

````python
articles = nyt.article_search(query = "Obama", results = 20) #query the articles
article = articles.copy() #turn object into dataframe 
print(article.head()) #view! 
````

#### `article_search()`: Filtering with `Dates`
To filter based on date range, you must import the package `datetime` in Python. That way, you can specify the timeframe in the range you want to pull articles from. `Dates` simply require a begin and an end.

````python
articles = nyt.article_search(query = "Obama", 
                              results = 20,
                              dates= {
                                  'begin': datetime.datetime(2008, 11, 1),
                                  'end': datetime.datetime(2008, 11, 30)
                              }) #query the articles
article = articles.copy() #turn object into dataframe 
print(article.head()) #view! 
````

#### `article_search()`: Filtering with `Options`
`Options` provide the most filter flexibility. Subfilters within options include: `sort`, `sources`, `news_desk`, `type_of_material`, `section_name`, `subject`, `body`, `headline`, and an advanced freeform filter query via `fq`. 

`sort` and `fq` filters must be inputted as a string value, whereas the remaining options require list format, regardless how the length. More on these subfilters - and identified issues - later on. 

````python
articles = nyt.article_search(query = "Obama", 
                              results = 20,
                              dates= {
                                  'begin': datetime.datetime(2008, 11, 1), #starting date
                                  'end': datetime.datetime(2008, 11, 30) #ending date 
                              },
                              options = {
                                  'sort': 'oldest', #begin from the start of articles in date frame 
                              }) #query the articles

article = articles.copy() #turn object into dataframe 
print(article.head()) #view! 
````

### How do I know what's causing a 'NoneType' error?
If you receive a TypeError: 'NoneType' object is not iterable, it's possible that either no articles were found with the query OR that the your query failed to access the API. The below code:
```python
resp = requests.get(
    "https://api.nytimes.com/svc/search/v2/articlesearch.json",
    params={
        "q": "KEYWORD",
        "api-key": "YOUR_KEY_HERE",
        "begin_date": "START_DATE",
        "end_date": "END_DATE",
    }
)
data = resp.json()
print(resp.status_code)
print(data.get("response", {}).get("meta", {}).get("hits"))
```
Since none of the subfilters function, you only need to input the query, begin_date, end_date, and your API key. 

Once ran, the first print statement will tell you the HTTP error code. A 200 status code means success! Hooray, your query retrieved the articles! Conversely, 400 means a bad request, so some type of issue in the query structure. 401 means unauthorized, likely an issue with your key itself, whereas 404 is a wrong endpoint. When you reach the rate limit, a 429 code will tell you that your key has made too many requests. 

The second, albeit a bit more complicatedly layer, mainly tells you the total "hits" or results that match your search query. If hits = 0, and there was NO error code present, then there were simply no articles to be found.
